```python
class Node:
  def __init__(self,data):
    self.data=data
    self.left=None
    self.right=None
def inorder(root):
  if root is None:
    return
  inorder(root.left)
  print(root.data,end="")
  inorder(root.right)
  #creating tree
root=Node(1)
root.left=Node(2)
root.right=Node(3)
root.left.left=Node(4)
root.left.right=Node(5)
inorder(root)
```

```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

visited = set()

def bfs(start):
    queue = deque([start])
    visited.add(start)

    while queue:
        node = queue.popleft()
        print(node, end=" ")

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)

print("BFS Traversal:")
bfs('A')
```

```python
def hill_climbing(values):
  current=0
  while True:
    left=current-1
    right=current+1
    if left>=0 and values[left]>values[current]:
      current=left
    elif right<len(values) and values[right]>values[current]:
      current=right
    else:
      return current,values[current]

values=[1,3,5,7,6,4,2]
position, maximum=hill_climbing(values)
print("Hill Climbing Result:")
print("Position:",position)
print("Maximum Value:",maximum)
```
